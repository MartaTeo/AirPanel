```python
import numpy as np
import re

humans['country_clean'] = humans['country'].str.strip().str.title()
humans = humans[humans['age'].notna() & (humans['age'].str.strip() != '')].copy()
humans.shape
```

```python
def classify_preference(text):
    if pd.isna(text) or str(text).strip() == '':
        return 'Non renseigné'
    t = text.lower()
    has_b = 'bouygues' in t
    has_o = 'orange' in t
    if has_b and not has_o:
        return 'Bouygues'
    if has_o and not has_b:
        return 'Orange'
    if has_b and has_o:
        m = re.search(r'préfér\w*\s[\w\s\']{0,15}(bouygues|orange)', t)
        if m:
            return 'Bouygues' if 'bouygues' in m.group() else 'Orange'
        return 'Ambiguous'
    if any(k in t for k in ['première', 'premiere', '1ère', 'la 1']):
        return 'Bouygues'
    if any(k in t for k in ['deuxième', 'deuxieme', 'seconde', 'la 2', '2ème']):
        return 'Orange'
    return 'Autre'

def classify_intent(text):
    if pd.isna(text) or str(text).strip() == '':
        return 'Non renseigné'
    t = text.lower().strip()
    if any(k in t for k in ['oui', 'tout à fait', 'certainement', 'absolument', 'pourquoi pas', 'au moins']):
        return 'Oui'
    if any(k in t for k in ['non', 'pas vraiment', 'pas du tout', 'peu probable', 'pas réellement']):
        return 'Non'
    if any(k in t for k in ['peut-être', 'possible', 'moyen', 'éventuellement', 'ni oui ni non', 'un peu']):
        return 'Peut-être'
    return 'Autre'

humans['pref_marque_cat']    = humans['Q_comparaison_1'].apply(classify_preference)
humans['pref_intention_cat'] = humans['Q_comparaison_5'].apply(classify_preference)
humans['intent_orange_cat']  = humans['Q_orange_intention_achat_1'].apply(classify_intent)
humans['intent_bouygues_cat']= humans['Q_bouygues_intention_achat_1'].apply(classify_intent)

humans[['pref_marque_cat','pref_intention_cat','intent_orange_cat','intent_bouygues_cat']].value_counts()
```

```python
import spacy

nlp = spacy.load('fr_core_news_md')
```

```python
text_cols = [
    'Q_bouygues_memorisation_1',
    'Q_bouygues_resonance_emotionnelle_1',
    'Q_bouygues_image_1',
    'Q_bouygues_intention_achat_1',
    'Q_orange_memorisation_1',
    'Q_orange_resonance_emotionnelle_1',
    'Q_orange_image_1',
    'Q_orange_intention_achat_1',
    'Q_comparaison_1',
    'Q_comparaison_3',
    'Q_comparaison_4',
    'Q_comparaison_5',
]

def extract_pos(text, pos_tag):
    if pd.isna(text) or str(text).strip() == '':
        return []
    doc = nlp(str(text))
    return [t.lemma_.lower() for t in doc
            if t.pos_ == pos_tag and not t.is_stop and not t.is_punct and len(t.lemma_) > 2]

for col in text_cols:
    humans[f'{col}_ADJ']  = humans[col].apply(lambda x: extract_pos(x, 'ADJ'))
    humans[f'{col}_NOUN'] = humans[col].apply(lambda x: extract_pos(x, 'NOUN'))
    humans[f'{col}_VERB'] = humans[col].apply(lambda x: extract_pos(x, 'VERB'))

humans.filter(regex='_ADJ|_NOUN|_VERB').head(3)
```

```python
from collections import Counter

def top_tokens(df, col_suffix, col_prefix, n=15):
    cols = [c for c in df.columns if c.startswith(col_prefix) and c.endswith(col_suffix)]
    tokens = [tok for col in cols for lst in df[col].dropna() for tok in lst if isinstance(lst, list)]
    return Counter(tokens).most_common(n)

results = {
    ('bouygues', '_ADJ'):  top_tokens(humans, '_ADJ',  'Q_bouygues'),
    ('bouygues', '_NOUN'): top_tokens(humans, '_NOUN', 'Q_bouygues'),
    ('bouygues', '_VERB'): top_tokens(humans, '_VERB', 'Q_bouygues'),
    ('orange',   '_ADJ'):  top_tokens(humans, '_ADJ',  'Q_orange'),
    ('orange',   '_NOUN'): top_tokens(humans, '_NOUN', 'Q_orange'),
    ('orange',   '_VERB'): top_tokens(humans, '_VERB', 'Q_orange'),
}

for (brand, pos), tokens in results.items():
    print(f"\n{brand.upper()} {pos}: {tokens}")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

specs = [
    (('bouygues', '_ADJ'),  'Bouygues — Adjectifs', '#1A6B96'),
    (('bouygues', '_NOUN'), 'Bouygues — Noms',       '#2E86AB'),
    (('bouygues', '_VERB'), 'Bouygues — Verbes',     '#48CAE4'),
    (('orange',   '_ADJ'),  'Orange — Adjectifs',   '#E76F00'),
    (('orange',   '_NOUN'), 'Orange — Noms',         '#F4A261'),
    (('orange',   '_VERB'), 'Orange — Verbes',       '#E9C46A'),
]

for ax, (key, title, color) in zip(axes.flatten(), specs):
    data = results[key]
    if data:
        words, counts = zip(*data)
        ax.barh(words[::-1], counts[::-1], color=color)
    ax.set_title(title, fontweight='bold', fontsize=10)
    ax.set_xlabel('Fréquence')

plt.suptitle('Analyse POS par marque', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('pos_by_brand.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
from wordcloud import WordCloud

def build_freq(df, col_prefix, suffixes=('_ADJ', '_NOUN')):
    tokens = []
    for suf in suffixes:
        cols = [c for c in df.columns if c.startswith(col_prefix) and c.endswith(suf)]
        tokens += [tok for col in cols for lst in df[col].dropna() for tok in lst if isinstance(lst, list)]
    return Counter(tokens)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
for ax, prefix, cmap, title in [
    (axes[0], 'Q_bouygues', 'Blues',   'Bouygues'),
    (axes[1], 'Q_orange',   'Oranges', 'Orange'),
]:
    freq = build_freq(humans, prefix)
    wc = WordCloud(width=700, height=350, background_color='white',
                   colormap=cmap, max_words=60).generate_from_frequencies(freq)
    ax.imshow(wc, interpolation='bilinear')
    ax.axis('off')
    ax.set_title(title, fontsize=13, fontweight='bold')

plt.suptitle('Image perçue des marques (Adj + Noms)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('wordclouds.png', dpi=150, bbox_inches='tight')
plt.show()
```
